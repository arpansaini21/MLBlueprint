#### Align equations at equal sign
$$
\begin{align}
a &= b + c \\
d &= e + f
\end{align}
$$



#### adding space between for more readibility
$$
\begin{align}
a\, &= \, b & \text{(very small space)} \newline
a \: &= \: b & \text{(small space)} \newline
a \; &= \; b & \text{(medium space)} \newline
a \quad &= \quad b & \text{(large space)} \newline
a \qquad &= \qquad b    & \text{(very large space)} \newline
\end{align}
$$
